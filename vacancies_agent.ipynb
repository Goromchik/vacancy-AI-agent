# Чтение и парсинг критериев с criteria.md

```python
try:
    text = open("criteria.md", encoding="utf-8").read()
except FileNotFoundError:
    print("Файл criteria.md не найден")
    log.error("criteria.md не найден")
    raise

search_block = re.search(r"## Ищу\n(.*?)(?=\n##|\Z)", text, re.DOTALL)
search_lines = re.findall(r"-\s*(.+)", search_block.group(1))

target_titles = []

for line in search_lines:
    line = line.strip().lower()

    main_title = re.sub(r"\s*\(.*?\)", "", line).strip()
    if main_title:
        target_titles.append(main_title)

    aliases = re.findall(r"\((.*?)\)", line)
    for alias in aliases:
        alias = alias.strip().lower()
        if alias:
            target_titles.append(alias)

target_titles = list(dict.fromkeys(target_titles))

skill_block = re.search(r"## Навыки\n(.*?)(?=\n##|\Z)", text, re.DOTALL)
skill_lines = re.findall(r"-\s*(.+)", skill_block.group(1))

my_skills = []
skill_aliases = {}

for line in skill_lines:
    variants = [x.strip().lower() for x in line.split("|")]
    canonical = variants[0]
    my_skills.append(canonical)
    for v in variants:
        skill_aliases[v] = canonical

my_skills = list(dict.fromkeys(my_skills))


salary_match = re.search(r"от\s*([\d\s]+)\s*рубл", text, re.IGNORECASE)
min_salary = int(salary_match.group(1).replace(" ", "")) if salary_match else 0

allowed_exp = []
exp_match = re.search(r"опыт:\s*([^\n]+)", text, re.IGNORECASE)
if exp_match:
    exp_line = exp_match.group(1).lower()
    raw_exps = re.split(r"\s*или\s*|\s*,\s*", exp_line)
    allowed_exp = [e.strip() for e in raw_exps if e.strip()]


city_match = re.search(r"город:\s*([^\n]+)", text, re.IGNORECASE)

preferred_city = "не указано"
allow_remote = False

if city_match:
    city_line = city_match.group(1).lower().strip()
    allow_remote = "удал" in city_line
    preferred_city = re.sub(r"\s*или\s+удал[её]нн[оая]\s*", "", city_line).strip()

print("Направления:", target_titles)
print("Город:", preferred_city)
print("Удалёнка допустима:", allow_remote)
print("Навыки:", my_skills)
print("Мин. зарплата:", min_salary)
print("Опыт:", allowed_exp)

log.info(f"Критерии загружены")
```

# Нормализация навыков

```python
def normalize_text(text):
    text = str(text).lower()
    text = text.replace("ё", "е")
    text = re.sub(r"[()]", "", text)

    for alias, canonical in sorted(skill_aliases.items(), key=lambda x: len(x[0]), reverse=True):
        text = text.replace(alias, canonical)

    return text
```

```python
try:
    df = pd.read_csv("vacancies.csv", encoding="utf-8-sig")
except FileNotFoundError:
    print("Файл vacancies.csv не найден")
    log.error("vacancies.csv не найден")

df["salary_min"] = pd.to_numeric(df["salary_min"], errors="coerce").fillna(0)
df["salary_max"] = pd.to_numeric(df["salary_max"], errors="coerce").fillna(0)
df["skills"] = df["skills"].fillna("")
df["remote"] = df["remote"].fillna("не указано")
df["experience"] = df["experience"].fillna("не указан")
df["location"] = df["location"].fillna("не указано")

df["skills_norm"] = df["skills"].apply(normalize_text)

print(f"Загружено вакансий: {len(df)}")
df.head()

log.info(f"Файл vacancies.csv загружен, строк: {len(df)}")
```

# Фильтрация по моим требованиям

```python
df = df[df["title"].str.lower().apply(lambda x: any(t in x for t in target_titles))]

if allowed_exp:
    df = df[df["experience"].isin(allowed_exp)]

df = df[(df["salary_min"] >= min_salary) | (df["salary_min"] == 0)]

if preferred_city != "не указано":
    if allow_remote:
        df = df[
            (df["remote"] == "да") |
            (df["location"].str.lower().str.contains(preferred_city, na=False))
        ]
    else:
        df = df[
            df["location"].str.lower().str.contains(preferred_city, na=False)
        ]

print(f"После фильтрации: {len(df)}")
log.info(f"Количество вакансий после фильтрации: {len(df)}")
```

```python
def score(row):
    points = 0
    skills_text = str(row["skills_norm"]).lower()
    title_text = str(row["title"]).lower()
    location_text = str(row["location"]).lower()

    matched = [skill for skill in my_skills if skill in skills_text]

    if any(t in title_text for t in target_titles):
        points += 4

    if "python" in matched:
        points += 3
    if "sql" in matched:
        points += 3

    for skill in matched:
        if skill not in ["python", "sql"]:
            points += 2

    if row["remote"] == "да":
        points += 4
    elif preferred_city != "не указано" and preferred_city in location_text:
        points += 2

    if row["salary_min"] >= min_salary and row["salary_min"] > 0:
        points += 3
    elif row["salary_min"] == 0:
        points -= 1

    # штраф за недостающие навыки из вакансии
    vacancy_skills = [s.strip() for s in skills_text.split(",") if s.strip()]
    lacking = [vs for vs in vacancy_skills if not any(my_s in vs for my_s in my_skills)]
    points -= min(len(lacking), 3)

    return points

if not df.empty:
    df["score"] = df.apply(score, axis=1)
    df = df.sort_values("score", ascending=False).reset_index(drop=True)

log.info("Скоринг завершён")
```

# Промт для LLM

```python
def make_summary(row):
    title_text = str(row.title).lower()
    vacancy_skills = [s.strip().lower() for s in str(row.skills_norm).split(",") if s.strip()]
    
    matched = [s for s in my_skills if s in row.skills_norm.lower()]

    lacking = []
    for vs in vacancy_skills:
        if vs and not any(my_s in vs for my_s in my_skills):
            lacking.append(vs)
    
    if any(t in title_text for t in target_titles):
        direction_status = "Подходит по направлению"
    else:
        direction_status = "Спорное совпадение по направлению"

    if row.salary_min >= min_salary:
        salary_status = "Подходит по зарплате"
    elif row.salary_min == 0:
        salary_status = "Зарплата не указана"
    else:
        salary_status = "Ниже ожиданий"

    if allow_remote and row.remote == "да":
        location_status = "Подходит по локации: есть удалёнка"
    elif preferred_city in str(row.location).lower():
        location_status = f"Подходит по локации: {row.location}"
    else:
        location_status = f"Не подходит по локации: {row.location}"

    prompt = f"""
Ты строгий карьерный консультант. Помогаешь студенту решить, подходит ли вакансия или нет.

Важные правила:
- Если "Удалёнка: нет" — не пиши про удалёнку в плюсах
- Если "Оценка локации" начинается со "Не подходит" — это минус, а не плюс
- Если "Оценка зарплаты: Ниже ожиданий" или "Зарплата не указана" — это минус, а не плюс

Данные кандидата:
- Умеет: {', '.join(my_skills)}
- Ищет зарплату от: {min_salary} руб
- Предпочтение по городу: {preferred_city}

Данные вакансии:
- Название: {row.title} ({row.company})
- Локация: {row.location}
- Оценка локации: {location_status}
- Зарплата: {row.salary_min} – {row.salary_max} руб. Оценка зарплаты: {salary_status}
- Удалёнка: {row.remote}
- Требуют навыки: {row.skills}
- Совпало: {', '.join(matched) if matched else 'ничего'}
- Кандидату не хватает: {', '.join(lacking) if lacking else 'хватает всего!'}

Пример 1:
Вход:
- Оценка локации: Подходит по локации: есть удалёнка
- Зарплата: Подходит по зарплате
- Совпало: python, sql, pandas
- Кандидату не хватает: airflow
Ответ:
Вердикт: Отлично подходит
Плюсы: Есть удалёнка, зарплата соответствует ожиданиям, хорошо совпадают навыки Python, SQL и pandas.
Минусы: Не хватает навыка Airflow.

Пример 2:
Вход:
- Оценка локации: Не подходит по локации: Москва
- Зарплата: Зарплата не указана
- Совпало: python, sql
- Кандидату не хватает: BI-инструменты
Ответ:
Вердикт: Спорно
Плюсы: Есть базовое совпадение по Python и SQL.
Минусы: Локация не совпадает с предпочтением кандидата, зарплата не указана, не хватает навыка BI-инструменты.

Напиши короткий анализ. Формат ответа строго такой:

Вердикт: <Отлично подходит / Подходит средне / Спорно>
Плюсы: <1 короткое предложение о совпадениях и условиях>
Минусы: <1 предложение о том, чего кандидату не хватает для этой роли, или про плохие условия>
"""
    try:
        response = client.chat.completions.create(
            model="llama-3.1-8b-instant",
            messages=[
                {"role": "system", "content": "Ты эксперт по найму. Не придумывай факты. Пиши коротко и только по шаблону. Если явных минусов нет, напиши 'Явных минусов нет'."},
                {"role": "user", "content": prompt}
            ],
            temperature=0.1
        )
        log.info(f"LLM summary создан для: {row.title}")
        return response.choices[0].message.content.strip()
    except Exception as e:
        log.warning(f"LLM недоступен для '{row.title}': {e}")
        return (
        f"LLM-объяснение не сгенерировано.\n"
    )
```

# Генерим отчёт

```python
lines = []
lines.append(f"# Отчёт по вакансиям — {datetime.now().strftime('%d.%m.%Y')}\n")

def format_salary(row):
    if row.salary_min > 0 and row.salary_max > 0:
        return f"{int(row.salary_min):,} – {int(row.salary_max):,} ₽".replace(",", " ")
    elif row.salary_min > 0:
        return f"от {int(row.salary_min):,} ₽".replace(",", " ")
    elif row.salary_max > 0:
        return f"до {int(row.salary_max):,} ₽".replace(",", " ")
    else:
        return "не указана"
    
if df.empty:
    lines.append("Подходящих вакансий не найдено")
    log.warning("После фильтрации не осталось вакансий")
else:
    for i, row in enumerate(df.head(5).itertuples(), start=1):
        print(f"Обрабатываю {i}: {row.title}")
        log.info(f"Обработка вакансии #{i}: {row.title}")
        summary = make_summary(row)

        lines.append(f"## #{i} {row.title} — {row.company}")
        lines.append(f"- Зарплата: {format_salary(row)}")
        lines.append(f"- Опыт: {row.experience}")
        lines.append(f"- Удалёнка: {row.remote}")
        lines.append(f"- Score: {row.score}")
        lines.append(f"- Ссылка: {row.url}")
        lines.append(f"\n{summary}\n")
        lines.append("")

with open("report.md", "w", encoding="utf-8") as f:
    f.write("\n".join(lines))
log.info("Отчёт сохранён в report.md")
```
